```python
orders_df = pd.DataFrame(sim.sales_orders)
lines_df = pd.DataFrame(sim.sales_order_lines)

header_totals = orders_df.set_index("sales_order_id")["order_total"]
line_totals = lines_df.groupby("sales_order_id")["extended_price"].sum()

max_difference = (
    header_totals.round(2) - line_totals.round(2)
).abs().max()

max_difference
```
