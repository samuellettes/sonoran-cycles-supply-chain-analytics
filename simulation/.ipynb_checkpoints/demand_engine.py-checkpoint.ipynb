```python
"""
demand_engine.py

Calculates daily demand for the Sonoran Cycles simulation.

Responsibilities
----------------
- Apply monthly seasonality
- Apply weekday effects
- Apply promotions
- Apply weather impacts
- Add random market variation
- Calculate demand indices for Dealer and DTC sales channels

The output of this module is a demand index that is used to
determine the number of dealer and direct-to-consumer orders
generated each day.
"""

import random
import numpy as np

import simulation.simulation_config as config

def get_monthly_multiplier(date):
    """
    Returns the demand multiplier for the month.
    """

    return config.MONTHLY_MULTIPLIER[date.month]

def get_weekday_multiplier(date, channel):
    """
    Returns weekday multiplier based on sales channel.
    """

    weekday = date.weekday()

    if channel == "Dealer":
        return config.DEALER_WEEKDAY[weekday]

    elif channel == "DTC":
        return config.DTC_WEEKDAY[weekday]

    else:
        raise ValueError("Channel must be 'Dealer' or 'DTC'")
        
def get_promotion_multiplier(date):
    """
    Returns promotion multiplier for a given day.
    """

    date_string = date.strftime("%Y-%m-%d")

    return config.PROMOTIONS.get(date_string, 1.0)

def get_weather_multiplier(weather):
    """
    Returns weather multiplier.
    """

    return config.WEATHER_MULTIPLIER.get(
        weather,
        1.0
    )

def get_random_noise():
    """
    Returns a random demand multiplier.
    """

    return np.random.normal(
        loc=config.NOISE_MEAN,
        scale=config.NOISE_STD_DEV
    )

def calculate_demand_index(
    date,
    channel,
    weather
):
    """
    Calculates total demand index.
    """

    demand_index = (

        get_monthly_multiplier(date)

        * get_weekday_multiplier(date, channel)

        * get_promotion_multiplier(date)

        * get_weather_multiplier(weather)

        * get_random_noise()

    )

    return round(demand_index, 3)

def calculate_daily_orders(
    date,
    channel,
    weather
):
    """
    Calculates number of orders to generate.
    """

    demand_index = calculate_demand_index(
        date,
        channel,
        weather
    )

    if channel == "Dealer":
        base_orders = config.BASE_DEALER_ORDERS

    elif channel == "DTC":
        base_orders = config.BASE_DTC_ORDERS

    else:
        raise ValueError("Invalid channel.")

    return max(
        0,
        round(base_orders * demand_index)
    )

def preview_day(date, weather):
    """
    Prints expected demand for a single day.
    """

    dealer = calculate_daily_orders(
        date,
        "Dealer",
        weather
    )

    dtc = calculate_daily_orders(
        date,
        "DTC",
        weather
    )

    print(f"Date: {date.date()}")
    print(f"Weather: {weather}")
    print(f"Dealer Orders: {dealer}")
    print(f"DTC Orders: {dtc}")
    
from datetime import datetime
from simulation.demand_engine import preview_day

preview_day(
    datetime(2023, 5, 15),
    "Sunny"
)

if channel == config.DEALER:
    ...

elif channel == config.DTC:
    ...
```
